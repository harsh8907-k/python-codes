```python
import pandas as pd

data ={
    "name":["nikhil","raj","harsh","pinu","pari","ragi"],
    "age":[23,45,45,78,89,78],
    "city":["benglore","rajkot","mumbai","delhi","haydrabad","surat"],
    "salary":[ 25000,50000,60000,20000,12000,52000]
    }
df = pd.DataFrame(data)
print(df)
print(df["salary"] > 30000)
print(df["age"] < 30)
print(df["city"] == "mumbai" )
print(df.sort_values("salary"))
df.head(1)
print(df.sort_values("age"))
df.tail(1)
df.info()
df.describe()
df.isnull()
df.dropna()
df.duplicated().sum()
df["salary"].mean()
```

```python
city_salary = df.groupby("city")["salary"].mean()
city_salary.plot(kind="bar")
plt.title("avarge salary by city")
plt.xlabel("city")
plt.ylabel("avrage salary")
plt.show()
```
